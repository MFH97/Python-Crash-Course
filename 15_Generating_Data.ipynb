# random walks

Using %%writefile magic command: You can create a new Python file directly in a Colab cell using the %%writefile magic command.

```python
%%writefile /content/random_walk.py
from random import choice

class RandomWalk:
    """A class to generate random walks."""

    def __init__(self, num_points=5000):
        """Initialize attributes of a walk."""
        self.num_points = num_points
        # All walks start at (0, 0)
        self.x_values = [0]
        self.y_values = [0]

    def fill_walk(self):
        """Calculate all the points in the walk."""

        # Keep taking steps until the walk reaches the desired length.
        while len(self.x_values) < self.num_points:
          # Decide which direction to go and how far to go in that direction
            x_direction = choice([1, -1])
            x_distance = choice([0, 1, 2, 3, 4])
            x_step = x_direction * x_distance

            y_direction = choice([1, -1])
            y_distance = choice([0, 1, 2, 3, 4])
            y_step = y_direction * y_distance

            # Reject moves that go nowhere
            if x_step == 0 and y_step == 0:
                continue
            # Calculate the next x and y values
            x = self.x_values[-1] + x_step
            y = self.y_values[-1] + y_step

            self.x_values.append(x)
            self.y_values.append(y)
```

**15-4. Modified Random Walks**: In the class RandomWalk, x_step and y_step are
generated from the same set of conditions . The direction is chosen randomly
from the list [1, -1] and the distance from the list [0, 1, 2, 3, 4] . Modify the
values in these lists to see what happens to the overall shape of your walks . Try
a longer list of choices for the distance, such as 0 through 8, or remove the −1
from the x or y direction list

```python
from random import choice
import matplotlib.pyplot as plt

class RandomWalk:
    """A class to generate random walks with customizable step patterns."""

    def __init__(self, num_points=5000,
                 directions=[-1, 1],
                 distances=[0, 1, 2, 3, 4]):
        """Initialize walk attributes with customizable step options."""
        self.num_points = num_points
        self.directions = directions
        self.distances = distances
        self.x_values = [0]
        self.y_values = [0]

    def fill_walk(self):
        """Calculate all points in the walk using current step settings."""
        while len(self.x_values) < self.num_points:
            # Get x and y steps (can be different patterns)
            x_step = self._get_step()
            y_step = self._get_step()

            # Reject moves that go nowhere
            if x_step == 0 and y_step == 0:
                continue

            # Calculate new position
            x = self.x_values[-1] + x_step
            y = self.y_values[-1] + y_step

            self.x_values.append(x)
            self.y_values.append(y)

    def _get_step(self):
        """Calculate a single step using current direction/distance settings."""
        direction = choice(self.directions)
        distance = choice(self.distances)
        return direction * distance

# Create comparison plots
plt.figure(figsize=(15, 10))

# Standard walk (original settings)
rw1 = RandomWalk(directions=[-1, 1], distances=[0, 1, 2, 3, 4])
rw1.fill_walk()
plt.subplot(2, 2, 1)
plt.plot(rw1.x_values, rw1.y_values, linewidth=1)
plt.title("Standard Walk\nDirections [-1,1], Distances [0-4]")
plt.axis('equal')

# Walk with longer distances
rw2 = RandomWalk(directions=[-1, 1], distances=list(range(0, 9)))
rw2.fill_walk()
plt.subplot(2, 2, 2)
plt.plot(rw2.x_values, rw2.y_values, linewidth=1)
plt.title("Longer Steps\nDirections [-1,1], Distances [0-8]")
plt.axis('equal')

# Walk with only positive x direction
rw3 = RandomWalk(directions=[1], distances=[0, 1, 2, 3, 4])
rw3.fill_walk()
plt.subplot(2, 2, 3)
plt.plot(rw3.x_values, rw3.y_values, linewidth=1)
plt.title("Positive X Only\nDirections [1], Distances [0-4]")
plt.axis('equal')

# Walk with asymmetric distances
rw4 = RandomWalk(directions=[-1, 1], distances=[0, 0.5, 1, 1.5, 2])
rw4.fill_walk()
plt.subplot(2, 2, 4)
plt.plot(rw4.x_values, rw4.y_values, linewidth=1)
plt.title("Smaller Steps\nDirections [-1,1], Distances [0,0.5,1,1.5,2]")
plt.axis('equal')

plt.tight_layout()
plt.show()
```

**15-5. Refactoring**: The method fill_walk() is lengthy . Create a new method
called get_step() to determine the direction and distance for each step, and
then calculate the step . You should end up with two calls to get_step() in
fill_walk():

```python
from random import choice

class RandomWalk:
    """A class to generate random walks."""

    def __init__(self, num_points=5000):
        """Initialize attributes of a walk."""
        self.num_points = num_points
        self.x_values = [0]
        self.y_values = [0]

    def get_step(self):
        """Calculate a single step's direction and distance."""
        direction = choice([1, -1])
        distance = choice([0, 1, 2, 3, 4])
        return direction * distance

    def fill_walk(self):
        """Calculate all the points in the walk."""
        while len(self.x_values) < self.num_points:
            x_step = self.get_step()
            y_step = self.get_step()

            # Reject moves that go nowhere
            if x_step == 0 and y_step == 0:
                continue

            # Calculate new position
            x = self.x_values[-1] + x_step
            y = self.y_values[-1] + y_step

            self.x_values.append(x)
            self.y_values.append(y)
```
